in this file we will try to implement linear regression from scratch

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
input_array = np.array([i for i in range(1,11)])
```

```python
print(input_array)
```

```python
output_array = np.array([i for i in range(10, 0 , -1)])
```

```python
print(output_array)
```

```python
plt.scatter(input_array, output_array)
plt.xlabel("input numbers")
plt.ylabel("output numbers")
plt.title("scatter plot")
plt.show()
```

now i have to initialise m and c

```python
m = 10
b = 10
```

```python
for i in range(1000):
    prediction_array = []
    for i in input_array:
        y_pred = m * i + b
        prediction_array.append(y_pred)

    loss_array = []
    for i in range(len(output_array)):#loss calculation
        loss_array.append((output_array[i] - prediction_array[i])**2)
    #now we will calculate total loss
    total_loss = 0
    for i in range(len(loss_array)):
        total_loss = total_loss + loss_array[i]
    print("the mean squared error is:", total_loss / len(input_array))
    #now here comes the gradient part
    learning_rate = 0.001

    gradient_b = 2*(m * np.mean(input_array) + b - np.mean(output_array))
    gradient_m = (2*(m * np.mean(input_array * input_array) + b * np.mean(input_array) - np.mean(input_array * output_array)))
    b = b - learning_rate * gradient_b
    m = m - learning_rate * gradient_m
```

```python
print("the final values of m and b are:" , m , b)
```

```python
final_output = []
for i in range(len(input_array)):
    final_output.append(m * input_array[i] + b)
print(final_output)
```
