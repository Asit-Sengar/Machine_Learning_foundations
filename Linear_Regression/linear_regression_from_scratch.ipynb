```python
output_array = [i for i in range(10, 0 , -1)]
```

now i have to initialise m and c

```python
m = 10
b = 10
```

```python
prediction_array = []
for i in input_array:
    y_pred = m * i + b
    prediction_array.append(y_pred)
```

```python
loss_array = []
for i in range(len(output_array)):
    loss_array.append((output_array[i] - prediction_array[i])**2)
```

```python
#now we will calculate total loss
total_loss = 0
for i in range(len(loss_array)):
    total_loss = total_loss + loss_array[i]
print("the mean squared error is:", total_loss / len(input_array))
```
