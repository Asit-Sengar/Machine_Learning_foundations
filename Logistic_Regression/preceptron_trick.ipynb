### Perceptron Trick Implementation
The Perceptron Trick learns a linear decision boundary by updating weights and bias based on prediction errors. For each sample, it computes z = w1*x1 + w2*x2 + b and predicts 1 if z > 0, else 0. Misclassified samples trigger weight updates: w = w + learning_rate * error * feature and b = b + learning_rate * error. This iterative process converges to a separating hyperplane.

```python
def perceptron_trick(X, Y):

    learning_rate = 0.01

    w1 = 0  # choosing random values
    w2 = 0  # choosing random values
    b = 0 # choosing random values

    epochs = 100

    for epoch in range(epochs):
        for i in range(len(X)):

            feature_1 = X[i, 0]
            feature_2 = X[i, 1]

            # calculation of z
            z = w1*feature_1 + w2 * feature_2 + b

            # predicting values
            if z > 0:
                Y_pred = 1
            else:
                Y_pred = 0

            # calculating error
            error = Y[i] - Y_pred

            # calculating new weights
            w1 = w1 + learning_rate * error * feature_1
            w2 = w2 + learning_rate * error * feature_2
            b = b + learning_rate * error

    return w1, w2, b
```

### Training the perceptron with the dataset

```python
#now calling the perceptron function
weight_1 , weight_2 , bias = perceptron_trick(X, Y)
```
