## Gradient Descent

Gradient descent is an optimization algorithm used to reduce the value of a loss function.
It works by repeatedly adjusting the parameter in the direction that lowers the error.
With each update, the model moves closer to a minimum point.

## Loss Function Behavior

The loss function is related to the value of $b$ and increases as the squared error grows.
In this example, the loss becomes larger when $b$ moves away from the optimal value.
This helps us understand why the algorithm tries to move toward the minimum.

## When the Slope Is Negative

If the slope is negative, the function is decreasing at that point.
To move toward the minimum, we increase the value of $b$ so the next step goes upward.
This adjustment helps the model reach a lower loss.

## When the Slope Is Positive

If the slope is positive, the function is increasing at that point.
To reach the minimum, we decrease the value of $b$ so the next step moves downward.
This helps the algorithm move in the correct direction.

## Update Rule

The new value of $b$ is updated using the formula $b_{new} = b_{old} - \eta \cdot \text{slope}$.
Here, $\eta$ is the learning rate, which controls how large each step is.
A smaller learning rate makes updates more careful, while a larger one makes them faster.

## Convergence Check

If the difference between the new and old value of $b$ becomes very small, the algorithm is nearing a solution.
This means the updates are no longer changing the parameter much.
At this stage, the model is close to the minimum of the loss function.
